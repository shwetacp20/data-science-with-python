numpy Fuctions

1. Aggregate functions

```python
import numpy as np
arr=np.array([1,2,3,4,5,6])
print("Sum",np.sum(arr))
print("Max",np.max(arr))
print("Min",np.min(arr))
print("comulatuve",np.cumsum(arr))
print("Cumulative Product:", np.cumprod(arr))

```

STATISTICAL FUNCTIONS

```python
import statistics as st
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([15, 25, 35, 45, 55])

print("Array 1:", arr1)
print("Array 2:", arr2)

print("Mean:", np.mean(arr1), np.mean(arr2))
print("Median:", np.median(arr1), np.median(arr2))
print("Mode:",st.mode(arr1),st.mode(arr2))
print("Standard Deviation:", np.std(arr1), np.std(arr2))
print("Variance:", np.var(arr1), np.var(arr2))
print("Minimum:", np.min(arr1), np.min(arr2))
print("Maximum:", np.max(arr1), np.max(arr2))
```
