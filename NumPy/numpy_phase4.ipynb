```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
print("Shape:", sales_data.shape)
print(sales_data[-3:])
print("Total sales every year:", np.sum(sales_data[:, 1:], axis=0))print("Total sales every year:", np.sum(sales_data[:, 1:], axis=0))
print("Minimum sales per restaurant:", np.min(sales_data[:, 1:], axis=1))
print("Maximum sales per year:", np.max(sales_data[:, 1:], axis=0))
print("Average sales per restaurant:", np.mean(sales_data[:, 1:], axis=1))


cumulative = np.cumsum(sales_data[:, 1:], axis=1)
print("Cumulative sales:\n", cumulative)
print("Average cumulative sales:", np.mean(cumulative, axis=0))

total_sales = np.sum(sales_data[:, 1:], axis=0)
monthly_average = total_sales / 12

print("Monthly average every year:", monthly_average)
```
