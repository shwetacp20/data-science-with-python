Arrays:
An array is defined as a collection of items that are stored in contiguous memory locations
* it is a container which can hold a fixed no of items and these items should be the same type
* A combination of arrays a lot of time.thye array can reduce the overall size of the code

creating array from scratch

```python
import numpy as np
#Zeros
zeros=np.zeros([2,3])
print("Zeros array\n",zeros)
#Ones
ones=np.ones([2,3])
print("ones array\n",ones)
#Full
full=np.full((2,3),7)
print("full array\n",full)
#random
random=np.random.random((2,2))
print("random array\n",random)
#Sequence
sequence = np.arange(1, 11)
print("Sequence\n", sequence)
```

vector,matrix and tensor

```python
import numpy as np
vector=np.array([1,2,3,4,5])
print("vector:",vector)

matrix=np.array([[1,2,3],
                [4,5,6]])
print("Matrix:\n",matrix)

tensor=np.array([[1,2],
                 [2,3],
                 [3,4]])
print("Tensor:\n",tensor)
```

Array properties

```python
print(f"Vector: {vector}\nShape: {vector.shape}\nDimensions: {vector.ndim}\nSize: {vector.size}\nData Type: {vector.dtype}")
print(f"Matrix:\n{matrix}\nShape: {matrix.shape}\nDimensions: {matrix.ndim}\nSize: {matrix.size}\nData Type: {matrix.dtype}")
print(f"Tensor:\n{tensor}\nShape: {tensor.shape}\nDimensions: {tensor.ndim}\nSize: {tensor.size}\nData Type: {tensor.dtype}")
```
