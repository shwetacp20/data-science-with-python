#### Section 1: Data Inspection

```python
"""Q1. Display the first 10 rows, the shape of the DataFrame, and the data types of each column."""
#-----------------------------Write Solution Code-----------------------------



import pandas as pd

df = pd.read_excel(r"C:\Users\91805\Downloads\Pandas\csec_attendance.xlsx")

# Display the first 10 rows
print("First 10 rows:")
print(df.head(10))

# Display the shape of the DataFrame
print("\nShape of the DataFrame:")
print(df.shape)

# Display the data types of each column
print("\nData types of each column:")
print(df.dtypes)


```

```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count. """
#-----------------------------Write Solution Code-----------------------------

import pandas as pd

# Check missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Identify columns containing null values
print("\nColumns with null values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# Check 'Never' values in Last Present Date
never_count = df["Last Present Date"].astype(str).str.strip().str.lower().eq("never").sum()

print("\nCount of 'Never' in Last Present Date:")
print(never_count)

# Total missing or 'Never' values in Last Present Date
missing_count = df["Last Present Date"].isnull().sum()

print("\nNull values in Last Present Date:")
print(missing_count)

print("\nTotal missing or 'Never' in Last Present Date:")
print(missing_count + never_count)
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""
#-----------------------------Write Solution Code-----------------------------
import re

# Rename columns to snake_case
df.columns = [
    re.sub(r'[^a-zA-Z0-9]+', '_', col.strip()).strip('_').lower()
    for col in df.columns
]

# Display the updated column list
print("Updated column names:")
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""
#-----------------------------Write Solution Code-----------------------------

# Convert attendance percentage from string to numeric
df["attendance"] = (
    df["attendance"]
    .astype(str)
    .str.replace("%", "", regex=False)
    .str.strip()
    .astype(float)
)

# Convert Last Present Date to datetime, handling "Never" as NaT
df["last_present_date"] = pd.to_datetime(
    df["last_present_date"].replace("Never", pd.NaT),
    errors="coerce"
)

# Display the updated columns and data types
print(df[["attendance", "last_present_date"]].head(10))

print("\nUpdated data types:")
print(df[["attendance", "last_present_date"]].dtypes)
```

#### Section 2: Filtering & Selection

```python
"""Q5. Filter and display all students whose Status is "Critical". Sort them by Attendance % in ascending order."""
#-----------------------------Write Solution Code-----------------------------

# Filter students with Critical status
critical_students = df[df["status"].str.strip().str.lower() == "critical"]

# Sort by attendance percentage in ascending order
critical_students = critical_students.sort_values(
    by="attendance",
    ascending=True
)

# Display all critical students
print("Students with Critical Status (sorted by attendance):")
print(critical_students)
```

```python
"""Q6. Extract all students who have 0 present days (fully absent). List their roll numbers, names, and consecutive absences."""
#-----------------------------Write Solution Code----------------------------

# Filter students who have 0 present days
fully_absent = df[df["present"] == 0]

# Display roll number, student name, and consecutive absences
result = fully_absent[
    ["roll_no", "student_name", "consecutive_absences"]
]

print("Students with 0 Present Days:")
print(result)
```

```python
"""Q7. Using .loc and .iloc, display:
(a) All "Safe" students using .loc
(b) The first 15 rows and columns 2-6 using .iloc """
#-----------------------------Write Solution Code-----------------------------
# (a) Display all "Safe" students using .loc
safe_students = df.loc[df["status"].str.strip().str.lower() == "safe"]

print("All Safe Students:")
print(safe_students)

# (b) Display the first 15 rows and columns 2-6 using .iloc
print("\nFirst 15 Rows and Columns 2-6:")
print(df.iloc[:15, 1:6])
```

```python
"""Q8. Find all students whose Consecutive Absences is greater than or equal to 5. Display their name, status, and last present date."""
#-----------------------------Write Solution Code-----------------------------

# Filter students with consecutive absences >= 5
students_absent = df[df["consecutive_absences"] >= 5]

# Display name, status, and last present date
result = students_absent[
    ["student_name", "status", "last_present_date"]
]

print("Students with Consecutive Absences >= 5:")
print(result)
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""

#-----------------------------Write Solution Code-----------------------------

# Group data by Status and calculate statistics
status_summary = df.groupby("status").agg(
    student_count=("status", "count"),
    average_attendance=("attendance", "mean"),
    max_consecutive_absences=("consecutive_absences", "max")
)

# Display the summary
print("Status-wise Student Summary:")
print(status_summary)
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
#-----------------------------Write Solution Code-----------------------------

import numpy as np

# Create Attendance Category
df["attendance_category"] = np.select(
    [
        df["attendance"] >= 90,
        (df["attendance"] >= 75) & (df["attendance"] < 90),
        (df["attendance"] >= 60) & (df["attendance"] < 75),
        df["attendance"] < 60
    ],
    ["Excellent", "Good", "Average", "Poor"],
    default="Unknown"
)

# Count students in each category using groupby()
category_count = df.groupby("attendance_category").size()

# Display the results
print("Student Count by Attendance Category:")
print(category_count)
```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""

#-----------------------------Write Solution Code-----------------------------

# Calculate totals using .agg()
totals = df.agg({
    "total_classes": "sum",
    "present": "sum",
    "absent": "sum"
})

# Display the results
print("Overall Attendance Statistics:")
print(totals)
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""

#-----------------------------Write Solution Code-----------------------------

# Group by status and calculate aggregate statistics
status_agg = df.groupby("status").agg(
    min_present=("present", "min"),
    max_present=("present", "max"),
    mean_present=("present", "mean"),
    total_absent=("absent", "sum")
)

# Display the results
print("Status-wise Attendance Aggregation:")
print(status_agg)
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""

#-----------------------------Write Solution Code-----------------------------

# Create the Risk Score column
df["risk_score"] = (
    (df["absent"] * 0.4) +
    (df["consecutive_absences"] * 0.6)
)

# Display the top 10 students with the highest risk score
top_10_risk = df.sort_values(
    by="risk_score",
    ascending=False
).head(10)

print("Top 10 Students with Highest Risk Score:")
print(top_10_risk)
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""

#-----------------------------Write Solution Code-----------------------------

# Extract numeric values from Classes Needed (75%)
df["classes_needed_numeric"] = (
    df["classes_needed_75"]
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
    .fillna("0")
    .astype(int)
)

# Display the original and new columns
print(df[["classes_needed_75", "classes_needed_numeric"]].head(10))
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """

#-----------------------------Write Solution Code-----------------------------

# Create Attendance Flag using apply() and lambda
df["attendance_flag"] = df["attendance"].apply(
    lambda x: "Defaulter" if x < 75 else "Safe"
)

# Display the result
print(df[["student_name", "attendance", "attendance_flag"]])
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""

#-----------------------------Write Solution Code-----------------------------

import pandas as pd

# Define attendance ranges and labels
bins = [0, 50, 75, 90, 100]
labels = ["Critical", "Warning", "Good", "Excellent"]

# Create attendance categories using pd.cut()
df["attendance_range"] = pd.cut(
    df["attendance"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Display the result
print(df[["student_name", "attendance", "attendance_range"]])
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""

#-----------------------------Write Solution Code-----------------------------

# Create pivot table showing student count
pivot_summary = pd.pivot_table(
    df,
    index="status",
    columns="attendance_category",
    aggfunc="size",
    fill_value=0
)

# Display the summary
print("Student Count by Status and Attendance Category:")
print(pivot_summary)
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""

#-----------------------------Write Solution Code-----------------------------

# Extract the base name (e.g., "ANANYA" from "ANANYA 1")
df["base_name"] = (
    df["student_name"]
    .astype(str)
    .str.strip()
    .str.extract(r"^([A-Za-z]+)", expand=False)
    .str.upper()
)

# Identify base names that appear more than once
duplicate_names = df[
    df["base_name"].duplicated(keep=False)
].sort_values("base_name")

# Display all rows with duplicate base names
print("Rows with Duplicate Base Names:")
print(duplicate_names)
```

```python
""""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""

#-----------------------------Write Solution Code-----------------------------

import pandas as pd

# Create a hypothetical Student_Details DataFrame
student_details = pd.DataFrame({
    "Roll No": df["roll_no"].unique(),
    "Branch": ["CSE"] * len(df["roll_no"].unique()),
    "Semester": [1] * len(df["roll_no"].unique())
})

# Rename Roll No to match the attendance DataFrame
student_details.rename(columns={"Roll No": "roll_no"}, inplace=True)

# Perform an inner join using Roll No
merged_df = pd.merge(
    df,
    student_details,
    on="roll_no",
    how="inner"
)

# Display the merged result
print("Merged Student Attendance and Details:")
print(merged_df)
```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""
# (a) Export all Critical students
critical_students = df[df["status"].str.strip().str.lower() == "critical"]
critical_students.to_csv("critical_students.csv", index=False)


# (b) Export summary statistics grouped by Status
status_summary = df.groupby("status").agg(
    student_count=("student_name", "count"),
    average_attendance=("attendance", "mean"),
    max_consecutive_absences=("consecutive_absences", "max")
)

status_summary.to_csv("status_summary.csv")


# (c) Export Top 10 students by Risk Score
df["risk_score"] = (
    df["absent"] * 0.4 +
    df["consecutive_absences"] * 0.6
)

top_10_risk = df.sort_values(
    by="risk_score", ascending=False
).head(10)

top_10_risk.to_csv("top_10_risk_students.csv", index=False)

print("All three CSV files exported successfully!")
```
