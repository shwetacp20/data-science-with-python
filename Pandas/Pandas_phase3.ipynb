```python
#Select columns
import pandas as pd

sdata = pd.read_csv(r"C:\datascience\datascience-with-python\Pandas\student.csv")
sdata[['firstName',"marks",'fees']]
#Select rows
sdata.loc[sdata.id==106]
sdata.loc[sdata.firstName=='Rohan']
#Select rows using condition
sdata.loc[sdata.age>20]
#Slicing rows
sdata.iloc[:6, :4]
sdata.iloc[2:7, 1:3]
```

```python
sdata["FullName"]=sdata["firstName"]+" "+sdata["lastName"]
sdata
```

Filter

```python
# Students scoring above 80
sdata[sdata['marks'] > 80]

```

```python
# Students aged 20
sdata[sdata['age'] == 20]

```

```python
# Students paying fees of 30000
sdata[sdata['fees'] == 30000]

```

```python
# Students scoring above 80 and paying fees below 30000
sdata[(sdata['marks'] > 80) & (sdata['fees'] < 30000)]

```

```python
# Select specific columns
sdata[['firstName', 'marks', 'fees']]
```

```python
sdata.where(sdata['marks'] > 80)
```

```python
sdata.loc[sdata['marks']>=85,"Result"]="Selected"
sdata.loc[sdata['marks']<85,"Result"]=" Not Selected"
sdata
```
